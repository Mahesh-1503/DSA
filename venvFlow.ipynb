# Windows + VS Code + Jupyter Setup Guide

A complete professional workflow for Python projects using **Virtual Environments**, **VS Code**, and **Jupyter Notebooks** on Windows.

---

# 1. Create a Project

```powershell
# Create project folder
mkdir my-project

# Move into project
cd my-project
```

Project structure:

```text
my-project/
```

---

# 2. Create a Virtual Environment

Create an isolated Python environment for the project:

```powershell
python -m venv .venv
```

Result:

```text
my-project/
│
├── .venv/
└── ...
```

---

# 3. Activate the Virtual Environment

```powershell
.\.venv\Scripts\Activate.ps1
```

Successful activation:

```text
(.venv) PS C:\...\my-project>
```

Verify that Python is coming from the virtual environment:

```powershell
python -c "import sys; print(sys.executable)"
```

Expected output:

```text
...\my-project\.venv\Scripts\python.exe
```

---

# 4. Upgrade pip

Always upgrade `pip` before installing packages:

```powershell
python -m pip install --upgrade pip
```

---

# 5. Install and Register a Jupyter Kernel

Install Jupyter kernel support:

```powershell
pip install ipykernel
```

Register the environment as a notebook kernel:

```powershell
python -m ipykernel install --user --name my-project --display-name "Python (my-project)"
```

This makes the environment available inside Jupyter and VS Code notebooks.

---

# 6. Install Common Packages

Core data science stack:

```powershell
pip install pandas numpy matplotlib
```

Optional packages:

```powershell
pip install requests
pip install scikit-learn
pip install openpyxl
```

---

# 7. Create `requirements.txt`

## Capture Exact Dependency Versions

```powershell
pip freeze > requirements.txt
```

View contents:

```powershell
type requirements.txt
```

Example:

```text
ipykernel==7.2.0
matplotlib==3.10.6
numpy==2.2.6
pandas==2.3.2
```

---

# 8. Install Dependencies from `requirements.txt`

When setting up the project on another machine:

```powershell
pip install -r requirements.txt
```

---

# 9. Open the Project in VS Code

```powershell
code .
```

---

# 10. Select the Python Interpreter

Open the Command Palette:

```text
Ctrl + Shift + P
```

Choose:

```text
Python: Select Interpreter
```

Select the project's virtual environment:

```text
...\my-project\.venv\Scripts\python.exe
```

---

# 11. Create a Jupyter Notebook

Create:

```text
practice.ipynb
```

Select kernel:

```text
Python (my-project)
```

Test the environment:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print(pd.__version__)
print(np.__version__)
```

---

# Daily Workflow

Each time you return to the project:

```powershell
cd my-project

.\.venv\Scripts\Activate.ps1

code .
```

---

# Deactivate the Virtual Environment

When you're finished working:

```powershell
deactivate
```

Before:

```text
(.venv) PS C:\...\my-project>
```

After:

```text
PS C:\...\my-project>
```

---

# Clone an Existing Project

If a teammate or another machine needs to set up the project:

```powershell
git clone <repo-url>

cd my-project

python -m venv .venv

.\.venv\Scripts\Activate.ps1

pip install -r requirements.txt

code .
```

# Quick Reference

```powershell
# Create project
mkdir my-project
cd my-project

# Create virtual environment
python -m venv .venv

# Activate virtual environment
.\.venv\Scripts\Activate.ps1

# Upgrade pip
python -m pip install --upgrade pip

# Install Jupyter kernel
pip install ipykernel

# Register notebook kernel
python -m ipykernel install --user --name my-project --display-name "Python (my-project)"

# Install common libraries
pip install pandas numpy matplotlib

# Save dependencies
pip freeze > requirements.txt

# Open VS Code
code .

# Deactivate environment
deactivate
```

# Version Control Best Practices

## Commit These Files

```text
requirements.txt
README.md
*.ipynb
*.py
```

## Do Not Commit

```text
.venv/
__pycache__/
.ipynb_checkpoints/
```

---

# Create a `.gitignore`

```gitignore
# Virtual environment
.venv/

# Python cache
__pycache__/

# Jupyter checkpoints
.ipynb_checkpoints/
```

---

# Recommended Development Workflow

1. Create a project directory.
2. Create and activate a virtual environment.
3. Install required packages.
4. Register a Jupyter kernel.
5. Develop notebooks and source code in VS Code.
6. Track dependencies with `requirements.txt`.
7. Commit source files and documentation to Git.
8. Exclude generated and environment-specific files using `.gitignore`.

This workflow provides a clean, reproducible, and professional Python development environment suitable for personal projects, data analysis, machine learning, and team collaboration.
