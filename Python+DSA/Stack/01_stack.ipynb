```python
class Node:
    def __init__(self, value):
        self.data = value
        self.next = None
```

```python
class Stack:
    def __init__(self):
        self.top = None
        
    def isempty(self):
        return self.top is None
    
    def push(self,value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node 
        
    def traverse(self):
        if self.isempty():
            print("Stack is empty")
            return
        current = self.top
        while current:
            print(current.data)
            current = current.next
            
    def peek(self):
        if self.isempty():
            print("Stack is empty")
        else:
            print("Top element is:", self.top.data)
    
    def pop(self):
        if self.isempty():
            print("Stack is empty")
        else:
            popped_value = self.top.data
            self.top = self.top.next
            # print("Popped element is:", popped_value) ## Uncomment this line if you want to print the popped value
            
    def __len__(self):
        count = 0
        current = self.top
        while current:
            count += 1
            current = current.next
        return count
```

- 1. Reverse a string using stack

```python
def reverse_string(text):
    print("Original string:", text)
    s=Stack()
    for char in text:
        s.push(char)
    reversed_text = ""
    while not s.isempty():
        reversed_text += s.top.data
        s.pop()
    return reversed_text
reverse_string("Hello, World!")
```

- Undo Redo

```python
def undoRedo(text, pattern):
    u = Stack()   # Undo stack
    r = Stack()   # Redo stack

    # Push all characters onto undo stack
    for char in text:
        u.push(char)

    # Process operations
    for char in pattern:
        if char == 'u':
            if not u.isempty():
                r.push(u.top.data)
                u.pop()

        elif char == 'r':
            if not r.isempty():
                u.push(r.top.data)
                r.pop()

    # Build result (reverse because stack pops from top)
    result = []
    while not u.isempty():
        result.append(u.top.data)
        u.pop()

    return ''.join(result[::-1])
```

```python
undoRedo("Hello, World!","uuuu")
```
