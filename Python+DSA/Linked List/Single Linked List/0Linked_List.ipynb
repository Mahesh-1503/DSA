```python
class Node:
    def __init__(self, value):
        self.data = value
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        ## Empty Linked List
        self.head = None
        ## No.of Nodes in the Linked List
        self.n = 0
    ## Adding length method to the LinkedList class
    def __len__(self):
        return self.n
    
    ## insert_at_beginning(head)
    def insert_head(self, value):
        ## new node
        new_node = Node(value)
        ## create connection
        new_node.next = self.head
        ## update head
        self.head = new_node
        ## update length
        self.n += 1
        
    ## Traversing the Linked List
    def __str__(self):
        ## start from head
        current = self.head
        result = ''
        ## loop until current is None
        while current is not None:
            result += str(current.data) + ' -> '
            current = current.next
        return result[:-3] # Remove the last arrow and space
    
    ## insert_at_end(tail)
    def append(self, value):
        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.n += 1
            return
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node
        self.n += 1
        
    ## insert_at_position
    def insert_after(self, after, value):
        new_node = Node(value)
        current = self.head
        while current is not None:
            if current.data == after:
                new_node.next = current.next
                current.next = new_node
                self.n += 1
                return
            current = current.next
        else:
            print(f"Node with value {after} not found.")
            
    ## Empty the Linked List
    def clear(self):
        self.head = None
        self.n = 0
        
    ## Delete head node
    def delete_head(self):
        if self.head is None:
            print("List is empty. Cannot delete head.")
            return
        self.head = self.head.next
        self.n -= 1
        
    ## Delete tail node
    def pop(self):
        if self.head is None:
            print("List is empty. Cannot delete tail.")
            return
        current = self.head
        if current.next is None:
            return self.delete_head()
        while current.next.next is not None:
            current = current.next
        current.next = None
        self.n -= 1
        
    ## Delete node by value
    def delete_by_value(self, value):
        if self.head is None:
            print("List is empty. Cannot delete node.")
            return
        if self.head.data == value:
            self.delete_head()
            return 
        current = self.head
        while current.next is not None:
            if current.next.data == value:
                current.next = current.next.next
                self.n -= 1
                return
            current = current.next
        print(f"Node with value {value} not found.")
        
    ## search for a node by value
    def search(self, value):
        current = self.head
        position = 0
        while current is not None:
            if current.data == value:
                return position
            current = current.next
            position += 1
        return 'Not Found'
    
    ## get item by index
    def __getitem__(self, index):
        if index < 0 or index >= self.n:
            return "Index out of bounds"
        current = self.head
        for _ in range(index):
            current = current.next
        return current.data
```

```python
L = LinkedList()
```

```python
L.insert_head(1)
L.insert_head(2)
L.insert_head(3)
L.insert_head(4)
```

##### Linked List:

- Insert [head, tail(append), middle(insert)]
- Traverse [print ] 
- Delete [head, tail(pop), value(remove), index]
- Search [value, index]
