# What is TLE Error in DSA?

TLE means:

# **TLE = Time Limit Exceeded**

It happens when:

> Your program takes **too much time** to run, so the online judge stops your program before it finishes.

This is one of the **most common errors** beginners face in Data Structures & Algorithms.

---

# Real-Life Analogy

Imagine:

Your teacher says:

> “Solve this problem in 2 minutes.”

But you take 10 minutes.

Even if your answer is correct…

❌ Teacher won’t wait forever.

Same in coding platforms like:

* [LeetCode](https://leetcode.com?utm_source=chatgpt.com)
* [Codeforces](https://codeforces.com?utm_source=chatgpt.com)
* [HackerRank](https://www.hackerrank.com?utm_source=chatgpt.com)
* [CodeChef](https://www.codechef.com?utm_source=chatgpt.com)

They give your code a fixed execution time like:

* 1 second
* 2 seconds
* 5 seconds

If your code exceeds that limit:

❌ **TLE Error**

---

# Why TLE Happens

Main reasons:

1. Bad Time Complexity
2. Unnecessary loops
3. Nested loops
4. Repeating calculations
5. Using brute force solution
6. Infinite loops
7. Large input size

---

# Most Important Beginner Understanding

## Small input → almost every code works

But:

## Large input → only optimized code works

Example:

If `n = 5`

Even slow code works.

But if:

`n = 100000`

Then inefficient code becomes extremely slow.

---

# Example 1 — Understanding TLE

Suppose problem says:

> Find sum from 1 to n

---

## Beginner Approach

```python
n = 100000000

total = 0

for i in range(1, n + 1):
    total += i

print(total)
```

---

# Time Complexity

Loop runs `n` times.

So:

O(n)

If `n = 100000000`

That means:

100 million iterations.

Very slow.

Possible TLE.

---

# Optimized Approach

Use formula:

\frac{n(n+1)}{2}

```python
n = 100000000

total = n * (n + 1) // 2

print(total)
```

---

# Time Complexity

Only one calculation.

O(1)

Extremely fast.

---

# Important Lesson

Two programs can give:

✅ Same answer

But:

* One gets Accepted
* One gets TLE

Because DSA is not only about correctness.

It is also about:

# Efficiency

Impossible within 1 second.

➡️ TLE

# Understanding Time Complexity vs TLE (Beginner Friendly)

When solving DSA problems, one question always matters:

> **“Will my code finish fast enough?”**

That is where **Time Complexity** comes in.

And when your program is too slow:

> You get **TLE (Time Limit Exceeded)**

---

# First Understand This

Suppose an online judge gives:

* `1 second` time limit
* Input size: `n = 100000`

Now imagine two solutions:

| Solution | Operations                |
| -------- | ------------------------- |
| O(n)     | 100,000 operations        |
| O(n²)    | 10,000,000,000 operations |

Second one is HUGE.

Even if both are “correct,” only the faster one passes.

---

# How “Good” or “Bad” Complexity is Decided

It depends on:

1. **Input Size (`n`)**
2. **Number of operations**
3. **Time limit**

---

# Approximate Competitive Programming Rule

| Operations | Usually Safe? |
| ---------- | ------------- |
| 10⁶        | Safe          |
| 10⁷        | Sometimes     |
| 10⁸        | Risky         |
| 10⁹+       | TLE likely    |

# Time Complexity Hierarchy

From BEST → WORST:

```text
O(1)
O(log n)
O(n)
O(n log n)
O(n²)
O(n³)
O(2ⁿ)
O(n!)
```

---

# 1. O(1) — Constant Time

## Meaning

No matter how large input becomes,
execution time stays almost SAME.

---

## Example 1 — Access Array Element

```python
arr = [10, 20, 30, 40]

print(arr[2])
```

Whether array size is:

* 10
* 1000
* 1 crore

Still accessing one index takes same time.

---

## Example 2 — Dictionary Lookup

```python
student = {"id": 101}

print(student["id"])
```

Dictionary lookup is near constant.

---

## Example 3 — Swap Two Variables

```python
a, b = b, a
```

Always same operations.

---

## Example 4 — Stack Push

```python
stack.append(5)
```

Single operation.

---

## Example 5 — Check First Character

```python
word[0]
```

Still one access.

---

# Why O(1) is Excellent

Because growth is almost zero.

Even huge inputs stay fast.

---

# How to Calculate Time Complexity

The biggest beginner mistake is:

> Looking at code and memorizing complexity instead of understanding **how operations grow**.

The goal is NOT:

```text
“How many operations exactly?”
```

The goal is:

```text
“How does the number of operations grow when n grows?”
```

---

# Step-by-Step Method

Whenever you see code:

## Step 1 — Identify input size

Usually:

```python
n = len(arr)
```

or

```python
input size = n
```

---

## Step 2 — Count how many times major operations run

Example:

```python
for i in range(n):
    print(i)
```

`print()` runs `n` times.

So:

```text
O(n)
```

---

# VERY IMPORTANT RULES

| Pattern             | Complexity |
| ------------------- | ---------- |
| Single operation    | O(1)       |
| One loop            | O(n)       |
| Nested loops        | O(n²)      |
| Triple loops        | O(n³)      |
| Divide by 2         | O(log n)   |
| Loop + divide       | O(n log n) |
| Recursive branching | O(2ⁿ)      |
| Permutations        | O(n!)      |

---

So complexity becomes:

```text
log₂ n
```

```text
O(log n)
```

```text
n work
+
log n divisions
```

```text
n × log n
```

```text
n times
```

```text
log n
```

```text
n × log n
```

# Final Complexity

```text
O(n log n)
```

# 5. O(n²) — Quadratic Time

# Example 1

```python
for i in range(n):
    for j in range(n):
        print(i, j)
```

---

# Step-by-Step

Outer loop:

```text
n times
```

For EACH outer iteration:

Inner loop runs:

```text
n times
```

Total:

```text
n × n
```

---

# Final Complexity

```text
O(n²)
```

---

# Example 2

```python
for i in range(n):
    for j in range(i, n):
        print(i, j)
```

Many beginners think this is different.

But total operations still roughly:

```text
n(n+1)/2
```

Ignore constants and lower terms.

So:

```text
O(n²)
```

---

# Important Big-O Rule

In Big-O:

## Ignore constants

```text
O(2n) → O(n)
O(5n) → O(n)
```

---

## Ignore lower-order terms

```text
O(n² + n) → O(n²)
O(n³ + n² + n) → O(n³)
```

Because large growth dominates.

---

# Final Complexity

```text
O(2ⁿ)
```

---

# Example 2 — Subsets

For every element:

```text
Take it
OR
Don't take it
```

Two choices per element.

For `n` elements:

```text
2 × 2 × 2 × ...
```

# Final Complexity

```text
O(2ⁿ)
```

# Most Important Interview Skill

# Multiplication vs Addition

---

# CASE 1 — Nested Loops → MULTIPLY

```python
for i in range(n):
    for j in range(n):
        pass
```

```text
n × n = O(n²)
```

---

# CASE 2 — Separate Loops → ADD

```python
for i in range(n):
    pass

for j in range(n):
    pass
```

```text
n + n = 2n
```

Ignore constant:

```text
O(n)
```
