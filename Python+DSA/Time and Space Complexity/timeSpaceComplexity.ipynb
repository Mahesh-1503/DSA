### Time Complexity

**Rate of increase in time with respect to input size.**

# Time Complexity — What It *Actually* Means

Many beginners think:

> “Time Complexity means how many seconds a program takes to run.”

That is **NOT fully correct.**

Time complexity is:

> **The rate at which execution time grows as input size grows.**

It studies **growth**, not exact clock time.

---

# Simple Real-Life Analogy

Imagine two people traveling 100 km:

* One uses a **bike**
* One uses a **sports car**

The sports car reaches faster.

But does that mean the **route itself** is better?

No.

The **distance growth** remains the same.

Similarly:

* Different computers run code at different speeds
* But the **algorithm growth pattern** remains same

---

# Why Actual Time Cannot Define Complexity

Same code can run differently on different machines.

Example:

| Machine          | Execution Time |
| ---------------- | -------------- |
| MacBook M3       | 5 sec          |
| Old Windows 7 PC | 15 sec         |

Does the algorithm change?

❌ No.

Only hardware changed.

That’s why we do NOT measure algorithms using exact seconds.

---

# What We Actually Measure

We measure:

> “How execution time increases when input size increases.”

---

# Example

Suppose a loop prints numbers.

```python
for i in range(n):
    print(i)
```

If:

* `n = 10` → 10 operations
* `n = 100` → 100 operations
* `n = 1000` → 1000 operations

The work grows **linearly** with input.

So complexity is:

```text
O(n)
```

Meaning:

> Time increases proportionally with input size.

---

# Key Insight

Time complexity ignores:

* CPU speed
* RAM
* Operating System
* Programming language optimizations
* Background applications

Because these vary from machine to machine.

Instead, complexity focuses on:

✅ Algorithm behavior
✅ Growth pattern
✅ Scalability

---

# Another Analogy — Restaurant Example

Imagine a waiter serving customers.

## Algorithm 1

Waiter checks every customer one by one.

If customers increase:

* 10 customers → 10 checks
* 100 customers → 100 checks

Growth is linear.

```text
O(n)
```

---

## Algorithm 2

Waiter uses a sorted booking system and halves search space each time.

* 100 customers → ~7 checks
* 1000 customers → ~10 checks

This grows very slowly.

```text
O(log n)
```

Even if both waiters work on different restaurants/computers:

* Growth pattern remains same

That’s what complexity measures.

---

# Important Understanding

Two programs may take:

* 1 second today
* 10 seconds tomorrow

depending on machine.

But if both grow like:

```text
n → 2n → 4n
time → 2x → 4x
```

then their complexity is still:

```text
O(n)
```

---

# Short Definition (Interview-Friendly)

> Time Complexity is the measurement of how the execution time of an algorithm grows with respect to input size.

---

# Beginner-Friendly Definition

> Time complexity does not measure exact execution time.
>
> It measures how fast the work increases when input size increases.

---

# Reflection Statement

A fast computer can make bad algorithms look fast for small inputs.

But when data becomes huge:

> Good algorithms win.

That is why companies care deeply about time complexity.


# Understanding Operations in Time Complexity

Consider this code:

```python
for i in range(1, 6):
    print("mahesh")
```

---

# Step-by-Step Breakdown

## Iterations

```python
range(1, 6)
```

generates:

```python
1, 2, 3, 4, 5
```

So loop runs:

```text
5 times
```

---

# What Happens in Each Iteration?

For every iteration, Python performs some work internally.

Simplified operations:

| Operation | Description               |
| --------- | ------------------------- |
| 1         | Get next value from range |
| 2         | Assign value to `i`       |
| 3         | Execute `print()`         |

So approximately:

```text
3 operations per iteration
```

---

# Total Operations

If loop runs 5 times:

```text
3 × 5 = 15 operations
```

---

# Generalizing with N

Suppose:

```python
for i in range(n):
    print("mahesh")
```

Now loop runs:

```text
N times
```

Each iteration still does approximately:

```text
3 operations
```

So total operations become:

```text
3 × N
```

or

```text
3N
```

---

# Important Concept

In Time Complexity:

We ignore constants.

So:

```text
3N
```

becomes:

```text
O(N)
```

because when input becomes huge, constant `3` does not matter much.

---

# Why Constants Are Ignored

Suppose:

| Input Size | 3N Operations | N Operations |
| ---------- | ------------- | ------------ |
| 10         | 30            | 10           |
| 1000       | 3000          | 1000         |
| 1,000,000  | 3,000,000     | 1,000,000    |

Growth pattern is still linear.

Both increase proportionally with `N`.

That’s why:

```text
O(3N) → O(N)
```

---

# Real Understanding

Time complexity is not about:

```text
exact number of operations
```

It is about:

```text
how operations grow as input grows
```

---

# Visual Intuition

```text
N = 10     → 30 operations
N = 100    → 300 operations
N = 1000   → 3000 operations
```

As input increases:

```text
operations increase linearly
```

Hence:

```text
O(N)
```

---

# Beginner-Friendly Conclusion

```python
for i in range(n):
    print("mahesh")
```

* Loop runs `N` times
* Each iteration performs constant work
* Total work becomes proportional to `N`

Therefore:

```text
Time Complexity = O(N)
```

---

# Golden Rule

When analyzing loops:

```text
Total Work =
(Work per iteration)
×
(Number of iterations)
```

That single idea solves most beginner time complexity problems.


# The 3 Golden Rules of Time Complexity

When calculating time complexity, beginners usually make 3 mistakes:

* Focusing on exact execution time
* Keeping constants
* Thinking about small inputs

Real algorithm analysis works differently.

---

# Rule 1 — Always Think About Worst Case

In interviews and real systems:

> We usually calculate complexity using the **worst-case scenario**.

Because software must survive even under heavy input.

---

# Real-Life Analogy — Traffic

Suppose someone asks:

> “How much time does it take to reach office?”

You don’t answer based on:

✅ Empty roads at 2 AM

You answer based on:

✅ Heavy traffic hours

Because that is the safer assumption.

Algorithms are analyzed the same way.

---

# Example 1 — Linear Search

```python id="p3kq8y"
arr = [10, 20, 30, 40, 50]

target = 50

for num in arr:
    if num == target:
        print("Found")
        break
```

---

## Best Case

If target is first element:

```text id="z71t7h"
1 comparison
```

Complexity:

```text id="x0z6xz"
Ω(1)
```

---

## Worst Case

If target is last element or absent:

```text id="j7wqho"
N comparisons
```

Complexity:

```text id="6nk0fj"
O(N)
```

---

# Why Worst Case Matters

Because systems should work reliably even when input becomes huge.

Google, Amazon, Netflix don’t design systems assuming “best case.”

They prepare for:

* Millions of users
* Heavy traffic
* Large datasets

---

# Rule 2 — Ignore Constant Values

Suppose:

```python id="hkk2xm"
for i in range(n):
    print(i)
    print(i)
    print(i)
```

Each iteration performs:

```text id="ukw86w"
3 operations
```

Total:

```text id="yjlwm0"
3N
```

But complexity becomes:

```text id="2qkjnv"
O(N)
```

NOT:

```text id="f0zlg9"
O(3N)
```

---

# Why Ignore Constants?

Because when input becomes huge:

| N         | N    | 3N   |
| --------- | ---- | ---- |
| 10        | 10   | 30   |
| 1000      | 1000 | 3000 |
| 1,000,000 | 1M   | 3M   |

The growth pattern is still linear.

Constant `3` does not fundamentally change scalability.

---

# Real-Life Analogy — Delivery Trucks

Imagine:

* Truck A carries 1 box per trip
* Truck B carries 3 boxes per trip

If packages increase massively:

Both trucks still make trips proportional to package count.

Growth behavior remains linear.

That’s why:

```text id="pp4ygv"
O(3N) → O(N)
```

---

# Rule 3 — Ignore Lower Order Terms

Suppose:

```python id="3m7e0u"
for i in range(n):
    print(i)

for j in range(n * n):
    print(j)
```

Operations:

```text id="4p2gri"
N + N²
```

Complexity becomes:

```text id="rm8o8n"
O(N²)
```

Why?

Because for very large inputs:

```text id="w2j4zi"
N² dominates N
```

---

# Example

| N    | N    | N²        |
| ---- | ---- | --------- |
| 10   | 10   | 100       |
| 100  | 100  | 10,000    |
| 1000 | 1000 | 1,000,000 |

Eventually smaller terms become irrelevant.

So:

```text id="4xj2m9"
O(N² + N) → O(N²)
```

---

# The Core Philosophy

Time complexity focuses on:

```text id="vln3m0"
How fast the algorithm grows
```

NOT:

* Exact milliseconds
* CPU speed
* Small optimizations
* Constant factors

---

# Types of Time Complexities

# 1. Big O Notation — Worst Case

Measures:

> Maximum possible growth

Most commonly used.

---

## Example

Linear search in worst case:

```text id="l4n0kn"
O(N)
```

---

# 2. Theta Notation — Average/Tight Bound

Measures:

> Typical or exact growth behavior

Represented as:

```text id="jzw9rw"
Θ
```

---

## Example

If an algorithm always performs roughly proportional work:

```text id="g1s2k8"
Θ(N)
```

---

# 3. Omega Notation — Best Case

Measures:

> Minimum possible time

Represented as:

```text id="ymxjlwm"
Ω
```

---

## Example

Searching first element immediately:

```text id="8rsgut"
Ω(1)
```

---

# Quick Comparison Table

| Notation | Meaning       | Focus       |
| -------- | ------------- | ----------- |
| Big O    | Worst Case    | Upper Bound |
| Theta    | Average/Exact | Tight Bound |
| Omega    | Best Case     | Lower Bound |

---

# Most Common Complexities

| Complexity | Name         | Fast/Slow |
| ---------- | ------------ | --------- |
| O(1)       | Constant     | Very Fast |
| O(log N)   | Logarithmic  | Excellent |
| O(N)       | Linear       | Good      |
| O(N log N) | Linearithmic | Efficient |
| O(N²)      | Quadratic    | Slow      |
| O(2ᴺ)      | Exponential  | Very Slow |

---

# How to Calculate Time Complexity

# Example 1 — Constant Time

```python id="42klmn"
a = 10
b = 20
print(a + b)
```

Operations do not depend on input.

Complexity:

```text id="byic7q"
O(1)
```

---

# Example 2 — Single Loop

```python id="v2w8rn"
for i in range(n):
    print(i)
```

Loop runs `N` times.

Complexity:

```text id="mz5rcc"
O(N)
```

---

# Example 3 — Nested Loop

```python id="ib22zl"
for i in range(n):
    for j in range(n):
        print(i, j)
```

Inner loop runs `N` times for every outer iteration.

Total:

```text id="4x4cxu"
N × N = N²
```

Complexity:

```text id="l0exf4"
O(N²)
```

---

# Example 4 — Loop Halving Input

```python id="yr22ha"
i = n

while i > 1:
    i = i // 2
```

Input reduces by half every iteration.

Pattern:

```text id="1gvlpx"
N → N/2 → N/4 → N/8
```

Complexity:

```text id="4ppjdi"
O(log N)
```

---

# Example 5 — Two Separate Loops

```python id="f9jlwm"
for i in range(n):
    print(i)

for j in range(n):
    print(j)
```

Operations:

```text id="4d11mt"
N + N = 2N
```

Ignore constant:

```text id="71gm4g"
O(N)
```

---

# Ultimate Cheat Sheet

| Pattern                      | Complexity            |
| ---------------------------- | --------------------- |
| Single loop                  | O(N)                  |
| Nested loops                 | O(N²)                 |
| Input halves every iteration | O(log N)              |
| Two independent loops        | Add complexities      |
| Nested dependent loops       | Multiply complexities |

---

# Final Intuition

Time complexity is basically asking:

> “If input becomes extremely huge…
>
> how badly will this algorithm suffer?”

That single question is the heart of algorithm analysis.


# Time Complexity of Sum of Natural Numbers

We want to calculate the time complexity of:

\frac{n(n+1)}{2}

---

# Step 1 — Understand the Formula

This formula calculates:

```text id="o6k4j1"
1 + 2 + 3 + 4 + ... + n
```

Example:

If:

```text id="ymn8tm"
n = 5
```

then:

```text id="6i5t8t"
1 + 2 + 3 + 4 + 5 = 15
```

Formula:

```text id="n7kh56"
5(5+1)/2 = 15
```

---

# Step 2 — Important Beginner Mistake

Many students see:

```text id="8kg1if"
n(n+1)/2
```

and think:

> “There is N here, so complexity must be O(N).”

❌ Wrong.

Time complexity is NOT about variables existing in formula.

It is about:

```text id="9obd3v"
How many operations the computer performs
```

---

# Step 3 — Count Operations

Suppose code is:

```python id="c9ysvq"
sum = n * (n + 1) // 2
```

Now count operations.

---

## Operation 1

```python id="jlwm2y"
n + 1
```

One addition.

---

## Operation 2

```python id="67nrrr"
n * (n + 1)
```

One multiplication.

---

## Operation 3

```python id="3bff26"
... // 2
```

One division.

---

# Total Operations

No matter how large `n` becomes:

* 1 addition
* 1 multiplication
* 1 division

Always fixed.

---

# Key Observation

If:

```text id="dqqm9d"
n = 10
```

operations = 3

If:

```text id="00m2gr"
n = 1,000,000
```

operations still = 3

If:

```text id="13g82j"
n = 1,000,000,000
```

operations still = 3

Operations are NOT growing with input.

---

# Step 4 — Remove Constants

Total operations:

```text id="e8mqvz"
3
```

Ignore constants in complexity analysis.

So:

```text id="f8jlwm"
O(3)
```

becomes:

```text id="9tfjlv"
O(1)
```

---

# Final Answer

The time complexity of:

\frac{n(n+1)}{2}

is:

```text id="q9b4zt"
O(1)
```

because the number of operations remains constant regardless of input size.

---

# Powerful Comparison

# Method 1 — Using Loop

```python id="74k5kr"
total = 0

for i in range(1, n + 1):
    total += i
```

Loop runs `N` times.

Complexity:

```text id="xjlwm9"
O(N)
```

---

# Method 2 — Using Formula

```python id="rgq8o8"
total = n * (n + 1) // 2
```

Only fixed operations.

Complexity:

```text id="2wlyv9"
O(1)
```

---

# Real Understanding

This is why formulas are powerful.

A loop suffers more as input grows.

But a direct mathematical formula does not.

---

# Real-Life Analogy

Imagine calculating salary of 1 crore employees.

## Loop Approach

Add every salary one by one.

Work increases massively.

---

## Formula Approach

Use one direct equation.

Same number of steps regardless of company size.

That is:

```text id="k2r6eu"
Constant Time Complexity
```

---

# Golden Rule to Remember

Ask yourself:

> “If input size increases enormously…
>
> will the number of operations also increase?”

* If YES → not constant
* If NO → `O(1)`

---

# Interview-Level Understanding

Even though the formula contains `N`:

\frac{n(n+1)}{2}

the computer does NOT iterate from `1` to `N`.

It simply performs a few arithmetic operations.

Hence:

```text id="i5qnj1"
Time Complexity = O(1)
```


# Space Complexity

# Space Complexity — Understanding Memory Usage

Most beginners focus only on:

```text id="8j7m2a"
Time Complexity
```

But efficient programs also care about:

```text id="r4k9zl"
Memory Usage
```

That is called:

# Space Complexity

---

# Simple Definition

> Space Complexity measures how much memory an algorithm uses with respect to input size.

Just like time complexity measures growth of time,

space complexity measures:

```text id="4m2xwb"
growth of memory usage
```

---

# Important Point

Space complexity is measured using:

```text id="sj9w2n"
Big O Notation
```

Examples:

| Memory Growth              | Complexity |
| -------------------------- | ---------- |
| Fixed memory               | O(1)       |
| Memory grows linearly      | O(N)       |
| Memory grows quadratically | O(N²)      |

---

# Space Complexity Formula

Space Complexity consists of:

\text{Space Complexity} = \text{Auxiliary Space} + \text{Input Space}

---

# 1. Input Space

Memory required to store input data.

Example:

```python id="s4mt8m"
arr = [10, 20, 30, 40]
```

Array itself occupies memory.

That memory is called:

```text id="n7xqz1"
Input Space
```

---

# 2. Auxiliary Space

Extra temporary memory used by algorithm.

Example:

```python id="sxy0bz"
total = 0
```

Variable `total` uses additional memory.

That is:

```text id="vjlwm7"
Auxiliary Space
```

---

# Real-Life Analogy

Imagine cooking.

## Input Space

Ingredients brought into kitchen:

* Rice
* Vegetables
* Oil

These already exist.

---

## Auxiliary Space

Extra things used while cooking:

* Bowls
* Spoons
* Plates

Temporary working memory.

---

# Example 1 — Constant Space

```python id="53y0tc"
a = 10
b = 20
c = a + b
```

---

## Memory Used

Variables:

* `a`
* `b`
* `c`

Fixed number of variables.

Even if input grows:

memory does NOT grow.

---

# Space Complexity

```text id="mjlwm0"
O(1)
```

Called:

```text id="a9kt2n"
Constant Space Complexity
```

---

# Example 2 — Linear Space

```python id="dg8jlwm"
arr = []

for i in range(n):
    arr.append(i)
```

---

# What Happens?

If:

```text id="r4x1fu"
N = 5
```

array stores 5 elements.

If:

```text id="c9v1ya"
N = 1000
```

array stores 1000 elements.

Memory grows with input size.

---

# Space Complexity

```text id="djlwm3"
O(N)
```

because memory usage increases linearly.

---

# Example 3 — Quadratic Space

```python id="7g2e1q"
matrix = []

for i in range(n):
    row = []
    
    for j in range(n):
        row.append(0)
        
    matrix.append(row)
```

---

# What Happens?

Creates:

```text id="shjlwm"
N × N matrix
```

If:

| N    | Elements  |
| ---- | --------- |
| 10   | 100       |
| 100  | 10,000    |
| 1000 | 1,000,000 |

Memory grows quadratically.

---

# Space Complexity

```text id="jjlwm2"
O(N²)
```

---

# Important Understanding

Space complexity also ignores:

* Constant memory
* Lower order terms

Just like time complexity.

---

# Example

```python id="5w7vlr"
arr = [0] * n
temp = [0] * n
x = 5
```

Memory:

```text id="g7jlwm"
N + N + 1
```

becomes:

```text id="d3jlwm"
2N + 1
```

Ignore constants:

```text id="u6jlwm"
O(N)
```

---

# Common Space Complexities

| Complexity | Meaning                      |
| ---------- | ---------------------------- |
| O(1)       | Constant memory              |
| O(N)       | Linear memory growth         |
| O(N²)      | Quadratic memory growth      |
| O(log N)   | Memory grows logarithmically |

---

# Difference Between Time & Space Complexity

| Time Complexity           | Space Complexity       |
| ------------------------- | ---------------------- |
| Measures execution growth | Measures memory growth |
| Focuses on operations     | Focuses on storage     |
| CPU-related               | RAM-related            |

---

# Real-World Insight

Sometimes developers trade:

* More memory for faster execution
  OR
* Less memory for slower execution

This is called:

```text id="6jlwm9"
Time-Space Tradeoff
```

---

# Example — Dictionary Lookup

Using dictionary:

```python id="jlwm4z"
data = {
    1: "A",
    2: "B"
}
```

Uses extra memory.

But searching becomes very fast.

So:

```text id="jlwm88"
More Space → Less Time
```

---

# Golden Rule

When calculating space complexity ask:

> “As input size grows…
>
> how much extra memory grows?”

That is the heart of space complexity.
