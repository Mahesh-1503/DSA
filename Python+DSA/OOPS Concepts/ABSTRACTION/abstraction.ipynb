```python
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    
    @abstractmethod
    def validate_secure_token(self):
        """Pure abstract method 1"""
        pass

    @abstractmethod
    def process_payment(self, amount):
        """Pure abstract method 2"""
        pass
```

```python
class PayPal(PaymentGateway):
    def validate_secure_token(self):
        print("PayPal: Validating OAuth2 token...")
        return True

    def process_payment(self, amount):
        print(f"PayPal: Successfully processed transaction of ${amount}.")

class Stripe(PaymentGateway):
    def validate_secure_token(self):
        print("Stripe: Checking API secret key...")
        return True

    def process_payment(self, amount):
        print(f"Stripe: Successfully charged ${amount} via credit card.")
```

```python
# The controller relies entirely on complete abstraction
def checkout_system(gateway: PaymentGateway, amount: float):
    print("--- Starting Checkout ---")
    if gateway.validate_secure_token():
        gateway.process_payment(amount)
    print("-------------------------\n")

if __name__ == "__main__":
    pp = PayPal()
    st = Stripe()

    # Executing through the pure abstracted interface
    checkout_system(pp, 100.50)
    checkout_system(st, 250.00)
```
