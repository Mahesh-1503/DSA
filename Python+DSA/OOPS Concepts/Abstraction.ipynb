#### Abstraction 

- Abstraction is the process of hiding the complex implementation details and showing only the necessary features of an object. It helps in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod
class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
class Laptop(Computer):
    # pass
    def process(self):
        return "Laptop is processing"
    
class Desktop(Computer):
    def process(self):
        return "Desktop is processing"
# Example usage
laptop = Laptop()
print(laptop.process())
desktop = Desktop()
print(desktop.process())
# Output:
# Laptop is processing
# Desktop is processing
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")

    def stop(self):
        print("Car stopped")

class Bike(Vehicle):
    def start(self):
        print("Bike started")

    def stop(self):
        print("Bike stopped")
# Example usage
car = Car()
car.start()
car.stop()

bike = Bike()
bike.start()
bike.stop()
# Output:
# Car started
# Car stopped
# Bike started
# Bike stopped
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def pay(self, amount, description):
        pass

    @abstractmethod
    def generate_receipt(self):
        pass
    
# Notice that Payment only defines what every payment method should do.

# It doesn't explain how.
```

```python
class CreditCardPayment(Payment):

    def authenticate(self):
        print("Verifying Card Number and CVV")

    def pay(self, amount, description):
        self.description=description
        self.amount=amount
        print(f"Paid {amount} using Credit Card")

    def generate_receipt(self):
        print(f"Credit Card Receipt Generated for: {self.amount}")
        print(f"{self.description}")
```

```python
class UPIPayment(Payment):

    def authenticate(self):
        print("Verifying UPI PIN")

    def pay(self, amount, description):
        self.description=description
        self.amount=amount
        print(f"Paid {amount} using UPI")

    def generate_receipt(self):
        print(f"UPI Receipt Generated for: {self.amount}")
        print(f"{self.description}")
        
    
```

```python
def make_payment(payment: Payment, amount: int, description: str):
    payment.authenticate()
    payment.pay(amount, description)
    payment.generate_receipt()
```

```python
credit = CreditCardPayment()
upi = UPIPayment()
```

```python
make_payment(credit, 5000, "Personal")
```

```python
make_payment(upi, 100, "Chicken")
```

```python
from abc import ABC, abstractmethod

# 1. The Abstract API Contract (The Interface)
class NotificationAPI(ABC):
    
    @abstractmethod
    def dispatch(self, recipient: str, payload: str) -> dict:
        """Every notification channel MUST implement this exact method signature."""
        pass

# 2. Concrete Implementation 1: Push Notification Microservice
class PushNotificationService(NotificationAPI):
    def dispatch(self, recipient: str, payload: str) -> dict:
        # Hidden complexity: Firebase Cloud Messaging (FCM) protocol, token validation, APNs payload formatting
        print(f"Connecting to APNs/FCM servers...")
        return {"status": "success", "channel": "push", "target": recipient}

# 3. Concrete Implementation 2: SMS Notification Microservice
class SMSNotificationService(NotificationAPI):
    def dispatch(self, recipient: str, payload: str) -> dict:
        # Hidden complexity: Twilio API gateway, carrier routing, SMS segment counting
        print(f"Routing via Twilio cellular gateway...")
        return {"status": "success", "channel": "sms", "target": recipient}

# --- Client Code / API Usage ---
def trigger_system_alert(service: NotificationAPI, user: str, message: str):
    # The system only knows about the abstract 'NotificationAPI' contract.
    # It doesn't care if it's Push or SMS!
    response = service.dispatch(user, message)
    print(f"Audit Log: Alert dispatched -> {response}\n")

# Execution
push_service = PushNotificationService()
sms_service = SMSNotificationService()

trigger_system_alert(push_service, "user_device_token_abc99", "Your package has shipped!")
trigger_system_alert(sms_service, "+1-555-0199", "Your verification code is 4821.")
```

```python
from abc import ABC, abstractmethod

# ==========================================
# 1. THE ABSTRACT API CONTRACT
# ==========================================
class SecurePaymentProcessor(ABC):
    
    @abstractmethod
    def authenticate_transaction(self) -> bool:
        """Enforces security verification without exposing internal secrets."""
        pass

    @abstractmethod
    def execute_transfer(self, amount_cents: int) -> str:
        """Executes the financial transfer and returns a transaction reference."""
        pass

    @abstractmethod
    def format_receipt(self, amount_cents: int, memo: str) -> dict:
        """Generates a standardized receipt payload."""
        pass

# ==========================================
# 2. CONCRETE IMPLEMENTATION: CREDIT CARD
# ==========================================
class CreditCardProcessor(SecurePaymentProcessor):
    def __init__(self, card_number: str, cvv: str):
        # Sensitive data is securely encapsulated (hidden) inside the object
        self.__card_number = card_number
        self.__cvv = cvv

    def authenticate_transaction(self) -> bool:
        # Hidden complexity: Luhn algorithm check, CVV regex validation
        print(f"[Secure Gateway] Validating Credit Card ending in {self.__card_number[-4:]}...")
        return len(self.__cvv) == 3

    def execute_transfer(self, amount_cents: int) -> str:
        print(f"[Credit Card] Charging ${amount_cents / 100:.2f} via Visa/Mastercard network...")
        return "TXN_CC_998123"

    def format_receipt(self, amount_cents: int, memo: str) -> dict:
        return {
            "gateway": "Credit Card",
            "amount_paid": f"${amount_cents / 100:.2f}",
            "description": memo,
            "status": "APPROVED"
        }

# ==========================================
# 3. CONCRETE IMPLEMENTATION: UPI
# ==========================================
class UPIProcessor(SecurePaymentProcessor):
    def __init__(self, vpa_id: str, pin: str):
        # Encapsulating UPI virtual address and private PIN
        self.__vpa_id = vpa_id
        self.__pin = pin

    def authenticate_transaction(self) -> bool:
        # Hidden complexity: NPCI gateway handshake, PIN hash verification
        print(f"[Secure Gateway] Authenticating VPA handle: {self.__vpa_id}...")
        return len(self.__pin) == 4

    def execute_transfer(self, amount_cents: int) -> str:
        print(f"[UPI] Transferring ${amount_cents / 100:.2f} via National Payments Corporation...")
        return "TXN_UPI_442109"

    def format_receipt(self, amount_cents: int, memo: str) -> dict:
        return {
            "gateway": "UPI",
            "amount_paid": f"${amount_cents / 100:.2f}",
            "description": memo,
            "status": "SUCCESS"
        }

# ==========================================
# 4. THE UNIFIED CHECKOUT PIPELINE (API CLIENT)
# ==========================================
def process_checkout(processor: SecurePaymentProcessor, amount_cents: int, memo: str):
    # The checkout system is completely abstracted. 
    # It doesn't know what kind of payment it is processing!
    
    if not processor.authenticate_transaction():
        print("❌ Payment Failed: Authentication rejected.")
        return

    transaction_id = processor.execute_transfer(amount_cents)
    
    receipt = processor.format_receipt(amount_cents, memo)
    print(f"\n--- Official Receipt [{transaction_id}] ---")
    for key, value in receipt.items():
        print(f"{key.capitalize()}: {value}")
    print("-" * 40 + "\n")

# ==========================================
# 5. EXECUTION
# ==========================================
# Instantiate concrete processors with their private credentials hidden inside
cc_processor = CreditCardProcessor("4111222233334444", "882")
upi_processor = UPIProcessor("alex@okhdfcbank", "9912")

# Run checkout polymorphically
process_checkout(cc_processor, 4999, "Cloud Server Hosting")
process_checkout(upi_processor, 1550, "Domain Name Registration")
```

```python
from abc import ABC, abstractmethod

# ==========================================
# 1. THE API CONTRACT (The Abstraction)
# ==========================================
class MusicPlayerAPI(ABC):
    
    @abstractmethod
    def play_track(self, song_title: str) -> None:
        """The public API endpoint every player MUST support."""
        pass

# ==========================================
# 2. HIDDEN COMPLEXITY: Spotify API
# ==========================================
class SpotifyPlayer(MusicPlayerAPI):
    def play_track(self, song_title: str) -> None:
        # Hidden complexity: OAuth tokens, Spotify Web API handshake, audio streaming buffers
        print(f"[Spotify API] Connecting to cloud servers... Now playing: {song_title}")

# ==========================================
# 3. HIDDEN COMPLEXITY: Apple Music API
# ==========================================
class AppleMusicPlayer(MusicPlayerAPI):
    def play_track(self, song_title: str) -> None:
        # Hidden complexity: iTunes DRM verification, AAC decoding protocols
        print(f"[Apple Music API] Unlocking DRM container... Now playing: {song_title}")

# ==========================================
# 4. HOW YOUR APP USES THE API
# ==========================================
def user_clicks_play(player: MusicPlayerAPI, song: str):
    # Your app ONLY knows about the abstract API. 
    # It doesn't care if it's Spotify or Apple Music!
    player.play_track(song)

# --- Execution ---
spotify = SpotifyPlayer()
apple = AppleMusicPlayer()

# The app calls the exact same abstracted API method for both
user_clicks_play(spotify, "Blinding Lights")
user_clicks_play(apple, "Midnight City")
```

```python
from abc import ABC, abstractmethod

# The blueprint for any toy that can speak
class TalkingToy(ABC):
   def __init__(self, name):
       self.name = name
   @abstractmethod
   def speak(self):
       pass

class RobotToy(TalkingToy):
   def speak(self):
       print(f'{self.name} says beep boop! I am a robot!')

class TeddyBearToy(TalkingToy):
   def speak(self):
       print(f"{self.name} says hug me! I'm cuddly!")

class DinosaurToy(TalkingToy):
   def speak(self):
       print(f'{self.name} says ROOOOAR!')

# Create toys
rusty = RobotToy('Rusty')
fluffy = TeddyBearToy('Fluffy')
rex = DinosaurToy('Rex')

toys = [rusty, fluffy, rex]
for toy in toys:
   toy.speak()

# Output:
# Rusty says beep boop! I am a robot!
# Fluffy says hug me! I'm cuddly!
# Rex says ROOOOAR!
```
