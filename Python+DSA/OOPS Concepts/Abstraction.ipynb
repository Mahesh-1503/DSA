#### Abstraction 

- Abstraction is the process of hiding the complex implementation details and showing only the necessary features of an object. It helps in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod
class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
class Laptop(Computer):
    # pass
    def process(self):
        return "Laptop is processing"
    
class Desktop(Computer):
    def process(self):
        return "Desktop is processing"
# Example usage
laptop = Laptop()
print(laptop.process())
desktop = Desktop()
print(desktop.process())
# Output:
# Laptop is processing
# Desktop is processing
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")

    def stop(self):
        print("Car stopped")

class Bike(Vehicle):
    def start(self):
        print("Bike started")

    def stop(self):
        print("Bike stopped")
# Example usage
car = Car()
car.start()
car.stop()

bike = Bike()
bike.start()
bike.stop()
# Output:
# Car started
# Car stopped
# Bike started
# Bike stopped
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def pay(self, amount, description):
        pass

    @abstractmethod
    def generate_receipt(self):
        pass
    
# Notice that Payment only defines what every payment method should do.

# It doesn't explain how.
```

```python
class CreditCardPayment(Payment):

    def authenticate(self):
        print("Verifying Card Number and CVV")

    def pay(self, amount, description):
        self.description=description
        self.amount=amount
        print(f"Paid {amount} using Credit Card")

    def generate_receipt(self):
        print(f"Credit Card Receipt Generated for: {self.amount}")
        print(f"{self.description}")
```

```python
class UPIPayment(Payment):

    def authenticate(self):
        print("Verifying UPI PIN")

    def pay(self, amount, description):
        self.description=description
        self.amount=amount
        print(f"Paid {amount} using UPI")

    def generate_receipt(self):
        print(f"UPI Receipt Generated for: {self.amount}")
        print(f"{self.description}")
```

```python
credit = CreditCardPayment()
upi = UPIPayment()
```

```python
def make_payment(payment: Payment, amount, description):
    payment.authenticate()
    payment.pay(amount, description)
    payment.generate_receipt()
```

## Another Example

```python
from abc import ABC, abstractmethod

# 1. The Abstract API Contract (The Interface)
class NotificationAPI(ABC):
    
    @abstractmethod
    def dispatch(self, recipient: str, payload: str) -> dict:
        """Every notification channel MUST implement this exact method signature."""
        pass

# 2. Concrete Implementation 1: Push Notification Microservice
class PushNotificationService(NotificationAPI):
    def dispatch(self, recipient: str, payload: str) -> dict:
        # Hidden complexity: Firebase Cloud Messaging (FCM) protocol, token validation, APNs payload formatting
        print(f"Connecting to APNs/FCM servers...")
        return {"status": "success", "channel": "push", "target": recipient}

# 3. Concrete Implementation 2: SMS Notification Microservice
class SMSNotificationService(NotificationAPI):
    def dispatch(self, recipient: str, payload: str) -> dict:
        # Hidden complexity: Twilio API gateway, carrier routing, SMS segment counting
        print(f"Routing via Twilio cellular gateway...")
        return {"status": "success", "channel": "sms", "target": recipient}

# --- Client Code / API Usage ---
def trigger_system_alert(service: NotificationAPI, user: str, message: str):
    # The system only knows about the abstract 'NotificationAPI' contract.
    # It doesn't care if it's Push or SMS!
    response = service.dispatch(user, message)
    print(f"Audit Log: Alert dispatched -> {response}\n")

# Execution
push_service = PushNotificationService()
sms_service = SMSNotificationService()

trigger_system_alert(push_service, "user_device_token_abc99", "Your package has shipped!")
trigger_system_alert(sms_service, "+1-555-0199", "Your verification code is 4821.")
```
