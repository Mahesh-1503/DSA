#### Encapsulation

- Encapsulation is one of the four fundamental OOP concepts. The other three are inheritance, polymorphism, and abstraction.

- Encapsulation is the process of wrapping the data (variables) and methods(functions) together as a single unit. It restricts direct access to some of the object's components. It is also known as data hiding. To prevent accidental data modification

- Encapsulation is implemented by using access specifiers. There are three types of access specifiers in python. They are: private, protected, public.

- Private: The data members and member functions defined as private are not accessible outside the class. They can only be accessed and modified by the member functions of the class.
- Protected: The data members and member functions defined as protected are accessible within the class and by classes derived from it. They cannot be accessed outside the class.

```python
## Encapsulation with Getter and Setter Methods
## Public, Private, and Protected Access Specifiers in Python

class Person:
    def __init__(self, name, age):
        self.name = name # public attribute
        self.__age = age # private attribute
    def get_age(self):
        return self.__age
person = Person("Ravi", 20)
## Accessing using get_age method
# print(person.get_age())
print(person.name)
print(person._Person__age) # accessing private attribute
person._Person__age = 30 # modifying private attribute
print(person._Person__age)

## Cant access private attribute directly
#print(person.__age) # This will raise an AttributeError
## Cant modify private attribute directly
#person.__age = 25 # This will create a new attribute, not modify the private one
print(person.__dict__) # to see all attributes of the object
# def get_age(self):
#     return self.__age
# print(person.get_age())
```

```python
## Protected: The data members and member functions defined as protected are accessible within the class and by classes derived from it. They cannot be accessed outside the class.

class Person:
    def __init__(self, name, age, gender):
        self.name = name # public attribute
        self._age = age # protected attribute
        self._gender = gender # protected attribute
class Employee(Person):
    def __init__(self, name, age, gender, employee_id):
        super().__init__(name, age, gender)
        self.employee_id = employee_id # public attribute
    def get_age(self):
        ## getter method to access protected attribute
        return self._age
employee = Employee("Ravi", 20, "Male", 12345)
print(employee.name) # accessing public attribute
print(employee.get_age()) # accessing protected attribute via method
print(employee._age) # accessing protected attribute directly
print(employee._gender) # accessing protected attribute directly
print(employee.employee_id) # accessing public attribute
```

#### Encapsulation with Getter and Setter Methods

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age ## private attribute
    def get_age(self):
        ## getter method to access private attribute
        return self.__age
    def set_age(self, age):
        ## setter method to modify private attribute
        if age > 0:
            self.__age = age
        else:
            print("Age cannot be negative")
person = Person("Ravi", 20)
print(person.get_age())
person.set_age(30)
print(person.get_age())
person.set_age(-5)
```
