````python
class Add:
    def __init__(self, a, b=0):
        self.a = a
        self.b = b
add1 = Add(2)
add2 = Add(5)
# Without operator overloading, this would raise an error
print(add1 * add2)  # TypeError: unsupported operand type(s) for +: 'Add' and 'Add'
````

````python
class Add:
    def __init__(self, a, b=0):
        self.a = a
        self.b = b
    def __add__(self, other): ## Operator Overloading for + here other is another object of class Add
        return Add(self.a + other.a, self.b + other.b)
    def __repr__(self):
        return f"Add({self.a}, {self.b})"
add1 = Add(2,3)
add2 = Add(5,7)
print(add1 + add2)  # Add object at 0x7f9b8c2d1d60
print((add1 + add2).a)  # 7
print((add1 + add2).b)  # 0
````

````python
class vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return vector(self.x + other.x, self.y + other.y)
    def __sub__(self, other):
        return vector(self.x - other.x, self.y - other.y)
    def __mul__(self, scalar):
        return vector(self.x * scalar, self.y * scalar)
    def __eq__(self, other): #__eq__ method to compare two vectors
        return self.x == other.x and self.y == other.y
    def __repr__(self): #__repr__ method for better visualization
        return f"{self.x} {self.y}"
v1 = vector(2, 3)
v2 = vector(5, 7)
print(v1 + v2)  # vector(7, 10)
print(v2 - v1)  # vector(3, 4)
print(v1 * 3)   # vector(6, 9)
print(v1 == v2) # False
print(v1 == vector(2, 3)) # True
````

## ⚙️ What Is Operator Overloading?

**Operator Overloading** means **giving special meaning to standard operators** (like `+`, `-`, `*`, `>`, etc.) **for user-defined objects**.

👉 In Python, operators like `+`, `-`, `==`, etc. are implemented using **special methods (dunder methods)** such as:

* `__add__` → for `+`
* `__sub__` → for `-`
* `__mul__` → for `*`
* `__gt__` → for `>`
* `__eq__` → for `==`
* `__str__` → for `print()`

## 🧩 Example 1: Basic Number Addition with Custom Class

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __str__(self):
        return str(self.value)

a = Number(10)
b = Number(20)
c = a + b   # calls a.__add__(b)
print(c)
```

✅ **Output:**

```
30
```

📘 Explanation:

* When you write `a + b`, Python automatically calls `a.__add__(b)`
* The `__add__` method defines what `+` means for your custom objects.

## ⚖️ Example 4: Comparison Operators

```python
class Box:
    def __init__(self, volume):
        self.volume = volume

    def __gt__(self, other):  # greater than
        return self.volume > other.volume

    def __lt__(self, other):  # less than
        return self.volume < other.volume

    def __eq__(self, other):  # equal to
        return self.volume == other.volume

b1 = Box(30)
b2 = Box(20)

print(b1 > b2)  # True
print(b1 < b2)  # False
print(b1 == b2) # False
```

💡 **Use case:** Compare custom objects (like files, boxes, containers) by size, weight, or value.

## 🧮 Example 7: Overloading `__len__` and `__str__`

```python
class BookShelf:
    def __init__(self, books):
        self.books = books

    def __len__(self):
        return len(self.books)

    def __str__(self):
        return ", ".join(self.books)

shelf = BookShelf(["Python", "Java", "C++"])
print(len(shelf))  # Calls __len__()
print(shelf)       # Calls __str__()
```

✅ **Output:**

```
3
Python, Java, C++
```
