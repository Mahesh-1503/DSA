## Inheritance:

Inheritance allows us to define a class that inherits all the methods and properties from another class. Parent class is the class being inherited from, also called base class. Child class is the class that inherits from another class, also called derived class.

  - Non private attributes
  - Non private methods

```python
class User:
    def __init__(self):
        self.name = "Mahesh"
        self.email = "mahesh@example.com"
        
        
    def login(self):
        print(f"Login Done")
        
    def register(self):
        print(f"Register Done")
        
class Student(User):
    def __init__(self):
        self.rollno = 12
        
    def enroll(self):
        print(f"Enrollment Done")
        
        
class Instructor(User):
    def __init__(self):
        self.idno = 50000
        
    def create(self):
        print(f"Instructor Created Course!")
```

```python
student = Student()
```

```python
student.enroll()
```

```python
student.login()
```
