1. Print 1 to 10 using recursion

````python
def printNumbers(i, n):
    """
    Recursively prints numbers from `i` to `n` (inclusive).

    This function uses recursion to print a sequence of integers in
    ascending order. It starts from the value of `i`, prints it, and
    then calls itself with the next number (`i + 1`) until it reaches
    a value greater than `n`.

    How it works:
    1. Check the base case:
       - If `i > n`, the recursion stops and the function returns.
    2. Print the current value of `i`.
    3. Recursively call `printNumbers(i + 1, n)` to print the next number.
    4. This process continues until the base case is met.

    Example:
        >>> printNumbers(1, 5)
        1
        2
        3
        4
        5

    Recursive Flow for printNumbers(1, 3):
        printNumbers(1, 3)
            ├─ prints 1
            └─ printNumbers(2, 3)
                   ├─ prints 2
                   └─ printNumbers(3, 3)
                          ├─ prints 3
                          └─ printNumbers(4, 3)
                                 └─ returns (base case)

    Parameters:
        i (int): The starting number to print.
        n (int): The ending number (inclusive).

    Returns:
        None: The function prints values directly and does not return
        any meaningful value.

    Time Complexity:
        O(n - i + 1)
        Each number in the range is processed exactly once.

    Space Complexity:
        O(n - i + 1)
        Due to the recursive call stack.
    """
    if i > n:
        return

    print(i)
    printNumbers(i + 1, n)
````

````python
printNumbers(1,5)
````

# Recursion Execution (`print_numbers(1, 5)`)

### Step-by-Step

1. `print_numbers(1, 5)` → prints `1` → calls `print_numbers(2, 5)`
2. `print_numbers(2, 5)` → prints `2` → calls `print_numbers(3, 5)`
3. `print_numbers(3, 5)` → prints `3` → calls `print_numbers(4, 5)`
4. `print_numbers(4, 5)` → prints `4` → calls `print_numbers(5, 5)`
5. `print_numbers(5, 5)` → prints `5` → calls `print_numbers(6, 5)`
6. `print_numbers(6, 5)` → base case reached → returns

### Call Stack

```text
print_numbers(1, 5)
└── print_numbers(2, 5)
    └── print_numbers(3, 5)
        └── print_numbers(4, 5)
            └── print_numbers(5, 5)
                └── print_numbers(6, 5)  # Base case
```

### Output

```python
1
2
3
4
5
```

### Key Points

* **Recursive Case:** Print `i`, then call `print_numbers(i + 1, n)`.
* **Base Case:** Stop when `i > n`.
* Since printing happens **before** the recursive call, numbers are printed in **ascending order**.

2. Factorial of a number using recursion

````python
def factorial(n):
    """Calculate factorial of n"""
    if n == 0:
        return 1
    return n * factorial(n - 1)
````

````python
print(factorial.__doc__)
print(factorial(5))
````

3. LeetCode 509 - Fibonacci Number

````python
def fib(n: int) -> int:
    """Calculate the nth Fibonacci number using recursion."""
    if n == 0 or n == 1:
        return n
    return fib(n - 1) + fib(n - 2)
````

````python
print(fib(4))
print(fib(5))
````

4. LeetCode 1137 - N-th Tribonacci Number

##### 📘 [Click here for explanation](./Tribonacci.pdf)🌈👈

````python
def tribonacci(n: int) -> int:
    """Calculate the nth tribonacci number using recursion."""
    if n == 0 or n == 1:
        return n
    if n == 2:
        return 1
    return tribonacci(n - 1) + tribonacci(n - 2) + tribonacci(n - 3)
````

````python
tribonacci(5)
````

````python
def isPowerOfTwo(n: int) -> bool:
    if n <= 0:
        return False
    if n == 1:
        return True
    if n % 2 != 0:
        return False
    return isPowerOfTwo(n // 2)
````

6. LeetCode 326 - Power of Three

````python
class Solution:
    def isPowerOfThree(self, n: int) -> bool:
        if n <= 0 or n % 3 != 0:
            return False
        if n == 1:
            return True
        return self.isPowerOfThree(n//3)
````

````python
obj = Solution()
print(obj.isPowerOfThree(45))
````

````python
## Without recursion
class Solution:
    def isPowerOfThree(self, n: int) -> bool:
        if n <= 0:
            return False
        
        while n % 3 == 0:
            n //= 3
            
        return n == 1
    
obj = Solution()
print(obj.isPowerOfThree(45))
````

### GCD AND LCM
---

````python
## Using Loop
def gcd(a,b):
    while b:
        a, b = b, a % b
    return a
gcd(48,18)
````

````python
## Using Recursion
def gcd(a,b):
    if b == 0:
        return a
    return gcd(b, a % b)
````

````python
gcd(48,18)
````

````python
def lcm(a, b):
    if b == 0:
        return a
    return a * b // gcd(a, b)
````

````python
lcm(48,18)
````

[7. LeetCode 50 - Pow(x, n)](./leetCode50.pdf) 👈Approach 1.

[7. LeetCode 50 - Pow(x, n)](./myPow(x,n)LeetCode50.pdf) 👇Approach 2.

````python
class Solution:
  def myPow(self, x: float, n: int) -> float:
    if n == 0:
      return 1
    if n < 0:
      return 1 / self.myPow(x, -n)
    if n % 2 == 1:
      return x * self.myPow(x, n - 1)
    return self.myPow(x * x, n // 2)
````

````python
solution = Solution()
print(solution.myPow(2.0, 10))
print(solution.myPow(2.0, -2))
````
