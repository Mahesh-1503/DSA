#### Inheritance

```python
## Parent class
class Car:
    def __init__(self, windows, doors, color, engineType):
        self.windows = windows
        self.doors = doors
        self.color = color
        self.engineType = engineType
    def displayInfo(self):
        return f"Car Info: {self.windows} windows, {self.doors} doors, Color: {self.color}, Engine Type: {self.engineType}"

## Child class
class Tesla(Car):
    def __init__(self, windows, doors, color, engineType, batteryCapacity, autopilot):
        super().__init__(windows, doors, color, engineType)
        self.batteryCapacity = batteryCapacity
        self.autopilot = autopilot
        
    def displayTeslaInfo(self):
        # parentInfo = super().displayInfo()
        return f"Windows: {self.windows}, Doors: {self.doors}, Color: {self.color}, Engine Type: {self.engineType}, Battery Capacity: {self.batteryCapacity} kWh, Autopilot: {self.autopilot}"

tesla1 = Tesla(4, 2, "Red", "Electric", 75, True)
teslaInfo = tesla1.displayTeslaInfo()
print(teslaInfo)

car1 = Car(4, 4, "Blue", "Gasoline")
carInfo = car1.displayInfo()
print(carInfo)
```

#### Multiple Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return f"{self.name} makes a sound."

class Pet:
    def __init__(self, owner):
        self.owner = owner
    def displayOwner(self):
        return f"Owner: {self.owner}"
    
class Dog(Animal, Pet):
    def __init__(self, name, owner, breed):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)
        self.breed = breed
    def speak(self):
        return f"{self.name} says Woof! Breed: {self.breed}"
dog1 = Dog("Buddy", "John", "Golden Retriever")
print(dog1.speak())  # Output: Buddy says Woof! Breed: Golden Retriever
print(dog1.displayOwner())  # Output: Owner: John
```

## Polymorphism

```python
## Method Overriding
## Base class
class Animal:
    def speak(self):
        return "Animal makes a sound."

class Dog(Animal):
    def speak(self):
        return "Dog barks."
class Cat(Animal):
    def speak(self):
        return "Cat meows."
    
###################################    
def display_sound(animal):
    print(animal.speak())
    
dog = Dog()
cat = Cat()
animal = Animal()

display_sound(dog)  # Output: Dog barks.
display_sound(cat)  # Output: Cat meows.
display_sound(animal)  # Output: Animal makes a sound.
```

```python
## Polymorphism
class Shape:
    def area(self):
        return  "The area of the shape"
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
    
    
    
#############
shape1 = Shape()
rect = Rectangle(5, 3)
circle = Circle(4)

print(shape1.area())  # Output: The area of the shape
print(rect.area())  # Output: 15
print(circle.area())  # Output: 50.24
```

## Encapsulation

```python
class person:
    def __init__(self, name, age):
        self.name = name  # public attribute
        self.__age = age    # private attribute
    def get_age(self):
        return self.__age
    

person1 = person("Ravi", 30)
print(person1.name)  # Output: Ravi
print(person1.get_age())  # Output: 30

```

```python
class person:
    def __init__(self, name, age, gender):
        self.name = name  # public attribute
        self.__age = age    # private attribute
        self._gender = gender  # protected attribute
class Employee(person):
    def __init__(self, name, age, gender, employee_id):
        super().__init__(name, age, gender)
        self.employee_id = employee_id
    def display_info(self):
        return f"Name: {self.name}, Gender: {self._gender}, Employee ID: {self.employee_id}"
employee1 = Employee("Alice", 28, "Female", "E123")
print(employee1.display_info())
```
